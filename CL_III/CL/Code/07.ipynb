```python
import numpy as np

# -------------------------------
# Step 1: Create Simple Data
# -------------------------------
# Healthy = near 0
healthy = np.array([[0.1, 0.2], [0.2, 0.1], [0.15, 0.25]])

# Damaged = higher values
damaged = np.array([[1.5, 1.6], [1.7, 1.8], [1.6, 1.5]])

# Combine data
X = np.vstack((healthy, damaged))
y = [0, 0, 0, 1, 1, 1]   # 0 = Healthy, 1 = Damaged

# -------------------------------
# Step 2: Generate Detectors
# -------------------------------
def generate_detectors(self_data):
    detectors = []
    for i in range(3):  # fixed small number
        d = np.random.uniform(1, 2, 2)  # far from healthy
        detectors.append(d)
    return np.array(detectors)

# -------------------------------
# Step 3: Classification
# -------------------------------
def classify(sample, detectors):
    for d in detectors:
        distance = np.linalg.norm(sample - d)
        if distance < 0.8:
            return 1   # Damaged
    return 0   # Healthy

# -------------------------------
# Step 4: Apply Model
# -------------------------------
detectors = generate_detectors(healthy)

predictions = []
for sample in X:
    predictions.append(classify(sample, detectors))

# -------------------------------
# Step 5: Accuracy Calculation
# -------------------------------
correct = 0
for i in range(len(y)):
    if y[i] == predictions[i]:
        correct += 1

accuracy = correct / len(y)

print("Actual Labels:     ", y)
print("Predicted Labels:  ", predictions)
print("Accuracy:", accuracy)
```

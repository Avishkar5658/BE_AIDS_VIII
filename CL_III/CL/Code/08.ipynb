```python
Implementation of DEAP (Distributed Evolutionary Algorithms in Python) Using
Python.
```

```python
import random
from deap import base, creator, tools

# -------------------------------
# Step 1: Define Fitness Function
# -------------------------------
# Minimize: f(x) = x^2
def evaluate(individual):
    return individual[0]**2,

# -------------------------------
# Step 2: Create Individual
# -------------------------------
creator.create("FitnessMin", base.Fitness, weights=(-1.0,))
creator.create("Individual", list, fitness=creator.FitnessMin)

toolbox = base.Toolbox()

# Attribute: random number between -10 to 10
toolbox.register("attr_float", random.uniform, -10, 10)

# Individual = 1 value
toolbox.register("individual", tools.initRepeat, creator.Individual, toolbox.attr_float, n=1)

# Population
toolbox.register("population", tools.initRepeat, list, toolbox.individual)

# Genetic operations
toolbox.register("evaluate", evaluate)
toolbox.register("mate", tools.cxBlend, alpha=0.5)
toolbox.register("mutate", tools.mutGaussian, mu=0, sigma=1, indpb=1.0)
toolbox.register("select", tools.selTournament, tournsize=3)

# -------------------------------
# Step 3: Initialize Population
# -------------------------------
population = toolbox.population(n=10)

# -------------------------------
# Step 4: Evolution Process
# -------------------------------
generations = 10

for gen in range(generations):
    # Evaluate fitness
    for ind in population:
        ind.fitness.values = toolbox.evaluate(ind)

    # Select
    offspring = toolbox.select(population, len(population))
    offspring = list(map(toolbox.clone, offspring))

    # Crossover
    for i in range(0, len(offspring)-1, 2):
        toolbox.mate(offspring[i], offspring[i+1])

    # Mutation
    for ind in offspring:
        toolbox.mutate(ind)

    population[:] = offspring

# -------------------------------
# Step 5: Best Solution
# -------------------------------
best = min(population, key=lambda ind: ind.fitness.values[0])

print("Best Solution (x):", round(best[0], 3))
print("Minimum Value f(x):", round(best[0]**2, 3))
```

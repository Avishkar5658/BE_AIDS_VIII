```python
 Implementation of Ant Colony Optimization for Solving the Traveling Salesman
Problem Using Python
```

```python
import numpy as np
import random

# -------------------------------
# Step 1: Define Cities (fixed)
# -------------------------------
cities = np.array([
    [0, 0],
    [1, 2],
    [4, 0],
    [5, 3]
])

num_cities = len(cities)

# -------------------------------
# Step 2: Distance Function
# -------------------------------
def distance(a, b):
    return np.linalg.norm(a - b)

# Distance matrix
dist = np.zeros((num_cities, num_cities))
for i in range(num_cities):
    for j in range(num_cities):
        dist[i][j] = distance(cities[i], cities[j])

# -------------------------------
# Step 3: ACO Parameters
# -------------------------------
pheromone = np.ones((num_cities, num_cities))
alpha = 1   # pheromone importance
beta = 2    # distance importance
iterations = 10
ants = 5

best_path = None
best_length = float('inf')

# -------------------------------
# Step 4: ACO Algorithm
# -------------------------------
for it in range(iterations):
    for ant in range(ants):
        path = [0]   # start from city 0
        visited = set(path)

        # Build path
        while len(path) < num_cities:
            current = path[-1]
            probs = []

            for j in range(num_cities):
                if j not in visited:
                    p = (pheromone[current][j]**alpha) * ((1/dist[current][j])**beta)
                else:
                    p = 0
                probs.append(p)

            probs = np.array(probs)
            probs = probs / probs.sum()

            next_city = np.random.choice(range(num_cities), p=probs)
            path.append(next_city)
            visited.add(next_city)

        path.append(0)  # return to start

        # Calculate distance
        length = 0
        for i in range(len(path)-1):
            length += dist[path[i]][path[i+1]]

        # Update best
        if length < best_length:
            best_length = length
            best_path = path

    # Update pheromone (simple)
    pheromone *= 0.9

# -------------------------------
# Step 5: Output
# -------------------------------
print("Best Path:", best_path)
print("Shortest Distance:", round(best_length, 2))
```

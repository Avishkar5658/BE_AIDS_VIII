```python
import numpy as np
import matplotlib.pyplot as plt

# Create fake images
content = np.ones((300,300,3)) * 200   # light image
style = np.ones((300,300,3)) * [255, 100, 100]  # reddish style

# Blend
output = (0.7 * content + 0.3 * style).astype(np.uint8)

# Show
plt.imshow(output)
plt.title("Stylized Output")
plt.axis('off')
plt.show()
```
